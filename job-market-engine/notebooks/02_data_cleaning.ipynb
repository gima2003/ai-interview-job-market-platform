```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import ast
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("../datasets/final_clean_jobs_dataset.csv")
df.head()
```

```python
df = df.drop_duplicates()
```

```python
df = df.dropna(subset=["job_title","company"])
```

```python
df["salary_max"] = df["salary_max"].fillna(0)
df["salary_min"] = df["salary_min"].fillna(0)
```

```python
df["job_title"] = df["job_title"].str.strip().str.title()
df["company"] = df["company"].str.title()
```

```python
df["salary_min"] = pd.to_numeric(df["salary_min"], errors="coerce")
df["salary_max"] = pd.to_numeric(df["salary_max"], errors="coerce")
```

```python
df["salary_min"] = df["salary_min"].fillna(0)
df["salary_max"] = df["salary_max"].fillna(0)
```

```python
df["salary_avg"] = (df["salary_min"] + df["salary_max"])/2
```

```python
df["skills"] = df["skills"].fillna("")
df["skills"] = df["skills"].str.lower()
df["skills"] = df["skills"].str.replace(";", ",")
```

```python
df["num_skills"] = df["skills"].apply(lambda x: len([s.strip() for s in x.split(",") if s.strip() != ""]))
```

```python
def get_job_level(title):
    title = str(title).lower()
    
    if "intern" in title:
        return "Intern"
    elif "junior" in title or "entry" in title:
        return "Junior"
    elif "senior" in title:
        return "Senior"
    else:
        return "Mid"

df["job_level"] = df["job_title"].apply(get_job_level)
```

```python
df["is_remote"] = df["location"].apply(
    lambda x: 1 if "remote" in str(x).lower() else 0
)
```

```python
df["experience_required"] = df["experience_required"].fillna("Not Specified")
df["experience_required"] = df["experience_required"].str.lower()
```

```python
df["degree_required"] = df["degree_required"].fillna("Not Specified")
df["degree_required"] = df["degree_required"].str.title()
```

```python
final_columns = [
    "job_id",
    "job_title",
    "company",
    "location",
    "salary_min",
    "salary_max",
    "description",
    "country",
    "search_keyword",
    "experience_required",
    "degree_required",
    "skills",
    "num_skills",
    "job_level",
    "is_remote",
    "salary_avg",
    "has_salary"
]

df = df[final_columns]
```

```python
df.to_csv("../datasets/final_clean_jobs_dataset.csv", index=False)
```
