```python
df = pd.read_csv("../datasets/final_clean_jobs_dataset.csv")
df.head()
```

```python
#Average Salary by Role
salary_by_role = (
    df.groupby("job_category")["salary_usd"]
    .mean()
    .round(0)
    .sort_values(ascending=False)
)

salary_by_role
```
