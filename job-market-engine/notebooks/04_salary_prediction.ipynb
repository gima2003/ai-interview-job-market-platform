```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.linear_model import LinearRegression

from sklearn.ensemble import RandomForestRegressor

import joblib
```

```python
df = pd.read_csv("../datasets/final_clean_jobs_dataset.csv")
df.head()
```

```python
df.columns
```

```python
#Average Salary by Role
salary_by_role = (
    df.groupby("job_category")["salary_usd"]
    .mean()
    .round(0)
    .sort_values(ascending=False)
)

salary_by_role
```

```python
salary_by_role.to_csv(
    "../outputs/salary_by_role.csv"
)
```

```python
# Salary Distribution chart
plt.figure(figsize=(10,6))
plt.hist(
    df["salary_usd"],
    bins=30
)
plt.title("Salary Distribution")

plt.xlabel("Salary USD")
plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    "../outputs/salary_distribution.png"
)

plt.show()
```

```python
country_salary = (
    df.groupby("company_location")["salary_usd"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

country_salary
```

```python
country_salary.plot(
    kind="bar",
    figsize=(10,6)
)

plt.title("Top Countries by Average Salary")

plt.tight_layout()

plt.savefig(
    "../outputs/salary_by_country.png"
)

plt.show()
```
