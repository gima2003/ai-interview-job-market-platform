```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("../datasets/jobs_raw.csv")
df.head()
```

```python
print("Rows: ", df.shape[0])
print("Columns: ", df.shape[1])
```

```python
df.columns
```

```python
df.info()
```

```python
missing = df.isnull().sum()

missing = missing[missing > 0]

missing
```

```python
duplicates = df.duplicated().sum()
print(
"Duplicate row: ",
    duplicates
    
)
```

```python
df = df.drop_duplicates(
    subset = "job_id"
)

df.shape
```

```python
df = df.dropna(
    subset = [
        "job_title",
        "skills"
    ]
)
```

```python
df.columns = (
    df.columns
    .str.lower()
    .str.replace(" ","_")
)
```

```python
df["job_title"] = (
    df["job_title"]
    .str.strip()
    .str.title()
)
```

```python
df["skills"] = (
    df["skills"]
    .astype(str)
    .str.lower()
    .str.strip()
)
```

```python
skill_replace = {
    "python programming": "python",
    "machine learning": "machine learning",
    "ml": "machine learning",
    "sql database": "sql"
}

def normalize_skills(skill_list):
    return [
        skill_replace.get(skill.lower(), skill.lower())
        for skill in skill_list
    ]

df["skills"] = df["skills"].apply(normalize_skills)
```

```python
df["skills"] = (
    df["skills"]
    .str.title()
)
```

```python
df[
[
"salary_min",
"salary_max",
"salary_avg"
]
].head()
```

```python
salary_columns = [
    "salary_min",
    "salary_max",
    "salary_avg"
]


for col in salary_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )
```

```python
df["salary_available"] = (
    df["salary_avg"]
    .notnull()
    .astype(int)
)
```

```python
df["salary_range"] = (

df["salary_max"]
-
df["salary_min"]

)
```

```python
df["salary_avg"].describe()
```

```python
df.info()
```

```python
df.to_csv(
"../datasets/clean_jobs.csv",
index=False
)
```
